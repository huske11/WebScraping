```python
conn = sqlite3.connect('lab_data.db')
cursor = conn.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS ria_page_info (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT,
    created_at TIMESTAMP
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS ria_news_titles (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    position INTEGER,
    title TEXT,
    created_at TIMESTAMP
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS ria_links (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    position INTEGER,
    link_text TEXT,
    url TEXT,
    created_at TIMESTAMP
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS ria_headers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    header_text TEXT,
    created_at TIMESTAMP
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS ria_texts (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    content TEXT,
    created_at TIMESTAMP
)
''')

conn.commit()
print("Таблицы созданы")
```

```python
now = datetime.now()

with open('bs4_data.txt', 'r', encoding='utf-8') as f:
    content = f.read()

import re
title_match = re.search(r'title: (.*?)\n', content)
if title_match:
    cursor.execute(
        "INSERT INTO ria_page_info (title, created_at) VALUES (?, ?)",
        (title_match.group(1), now)
    )

news_section = re.search(r'news_titles:\n(.*?)\n\n', content, re.DOTALL)
if news_section:
    news_titles = news_section.group(1).strip().split('\n')
    for pos, title in enumerate(news_titles, 1):
        if title.strip():
            cursor.execute(
                "INSERT INTO ria_news_titles (position, title, created_at) VALUES (?, ?, ?)",
                (pos, title.strip(), now)
            )

print("Данные из BS4 внесены в таблицу")
```

```python
with open('xpath_data.txt', 'r', encoding='utf-8') as f:
    xpath_content = f.read()

headers_section = re.search(r'headers:\n(.*?)\n\n', xpath_content, re.DOTALL)
if headers_section:
    headers = headers_section.group(1).strip().split('\n')
    for header in headers:
        if header.strip():
            cursor.execute(
                "INSERT INTO ria_headers (header_text, created_at) VALUES (?, ?)",
                (header.strip(), now)
            )

texts_section = re.search(r'texts:\n(.*)', xpath_content, re.DOTALL)
if texts_section:
    texts = texts_section.group(1).strip().split('\n')
    for text in texts:
        if text.strip():
            cursor.execute(
                "INSERT INTO ria_texts (content, created_at) VALUES (?, ?)",
                (text.strip(), now)
            )

conn.commit()
print("Данные из XPath внесены в таблицу")
```

```python
# Проверка
print("Статистика:")
tables = ['ria_page_info', 'ria_news_titles', 'ria_links', 'ria_headers', 'ria_texts']
for table in tables:
    cursor.execute(f"SELECT COUNT(*) FROM {table}")
    count = cursor.fetchone()[0]
    print(f"{table}: {count} записей")
```
