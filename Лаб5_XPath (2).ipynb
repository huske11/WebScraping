```python
import requests
from lxml import html
```

```python
url = 'https://ria.ru/'

response = requests.get(url)

tree = html.fromstring(response.content)

headers_xpath = '//h1/text() | //h2/text() | //h3/text() | //h4/text() | //h5/text() | //h6/text()'

text_xpath = '//text()[normalize-space()]'

headers = tree.xpath(headers_xpath)

text_elements = tree.xpath(text_xpath)

print("Заголовок:")
for header in headers:
    print(header.strip())

print("Текст:")
for text in text_elements:
    print(text.strip())
```

```python
response = requests.get(url)
tree = html.fromstring(response.content)

headers_xpath = '//h1/text() | //h2/text() | //h3/text() | //h4/text() | //h5/text() | //h6/text()'
headers = tree.xpath(headers_xpath)

visible_text_xpath = '//body//*[not(self::script) and not(self::style) and not(self::noscript)]/text()[normalize-space()]'
text_elements = tree.xpath(visible_text_xpath)

print("Заголовки:")
unique_headers = []
for header in headers:
    cleaned = header.strip()
    if cleaned and cleaned not in unique_headers:
        unique_headers.append(cleaned)
        print(cleaned)

print("\nТекст страницы:")
unique_text = []
for text in text_elements:
    cleaned = text.strip()
    if (len(cleaned) > 1 and
        not cleaned.startswith(('var ', 'function', 'BX.', '$(', 'if(', 'new ', 'ym(')) and
        not cleaned.startswith(('/bitrix/', '{"', '//')) and
        cleaned not in unique_text):
        unique_text.append(cleaned)
        print(cleaned)
```
